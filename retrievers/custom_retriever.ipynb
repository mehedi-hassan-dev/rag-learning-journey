```python
from typing import List

from langchain_core.documents import Document
from langchain_core.retrievers import BaseRetriever
```

```python
# Custom retriever বানাতে BaseRetriever inherit করতে হবে
class SimpleKeywordRetriever(BaseRetriever):
   
    documents: List[Document]   # Pydantic field হিসেবে documents ডিক্লেয়ার করা হলো

    # এই method-টাই মূল retrieval logic implement করে
    def _get_relevant_documents(
        self,
        query: str,
        **kwargs,
    ) -> List[Document]:
        results = []

        # Query-কে ছোট keyword-এ ভাগ করছি
        keywords = query.lower().split()

        # প্রতিটি document-এ keyword আছে কি না check করছি
        for doc in self.documents:
            content = doc.page_content.lower()

            # যেকোনো একটি keyword match করলে result-এ রাখব
            if any(keyword in content for keyword in keywords):
                results.append(doc)

        return results
```

```python
# Sample documents
documents = [
    Document(page_content="LangChain is used for LLM applications."),
    Document(page_content="Retrievers return relevant documents."),
    Document(page_content="Chroma is a vector database."),
]

# Custom retriever তৈরি করছি (keyword argument দিয়ে)
retriever = SimpleKeywordRetriever(documents=documents)

query = "vector database"

results = retriever.invoke(query)

print("Query:", query)
print("-" * 50)

for i, doc in enumerate(results, start=1):
    print(f"{i}. {doc.page_content}")
```
